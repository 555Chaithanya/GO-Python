Conditional Statements
in
Python programming

```python
x=int(input("enter a num:"))
print("Hello")
if x<10:
  print("wahoo!")
  print("wahoo!")
elif x<=20:
  print("hoohoo!")
  print("hoohoo!")
else:
  print("boohoo!")
  print("boohoo!")
print("goodbye")
```

```python
x=int(input("enter a num:"))
print("Hello")
if x>10:              #only if is executed, rest ignored called "Short-Circuit Evaluation"
  print("wahoo!")
  print("wahoo!")
elif x<=20:
  print("hoohoo!")
  print("hoohoo!")
else:
  print("boohoo!")
  print("boohoo!")
print("goodbye")
```

```python
x=int(input("enter a num:"))
print("Hello")
if x<10:
  print("wahoo!")
  print("wahoo!")
else:
  if x<=20:
    print("hoohoo!")
    print("hoohoo!")
  else:
    print("boohoo!")
    print("boohoo!")
print("goodbye")
```

expression evaluates to False
• "or" statements short circuit as soon as first
