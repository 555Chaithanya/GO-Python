Default Arguments:

```python
def greeting(name='friend'):
  print("Hello,",name)
greeting("Mike")
greeting()#takes default argument
```

```python
def greeting(name):
  print("Hello,",name)
greeting("Mike")
greeting()
```

```python
def foo(x,y=10):
  print(x+y)
foo(5)
foo(5,100)
```

```python
def foo(x=20,y=10):
  print(x+y)
foo(5)
foo()
foo(5,100)
```

```python
def foo(x,y=0):
  return x,y
result = foo(5)
print(result)
```

```python
def say(s,times=1):
  print(s*times)
say("Hello")
say("World ",3)
```
