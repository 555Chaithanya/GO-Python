```python
xList = [10, 20, 30, 40, 50, 60]
yList = []
zList = []
for j in range(1, 7, 2):  # j will be 1 and 3 and 5
    xList[j] = j
    yList.append(j)
    zList.append(yList)  #all elements in zList referencing the same list


print("zList:", zList)   #use visualiser to know what is happening
#https://cs1110.cs.cornell.edu/tutor/
```
