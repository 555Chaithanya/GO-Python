continue skips certain number given in sequence

```python
for i in range(1,11):
  if i==5:            #5 is skipped
    continue
  print(i)
```

```python
#print all squares of elements given list skipping list elements >10
num=[2,3,11,7,5,20,4]
for q in num:
  if q>10:
    continue
  print("Current number is:",q)
  sqr=q*q
  print("Square of ",q,"is",sqr)
```

List Comprehensions

Python provides an elegant mechanism for building a list by embedding
a for within list brackets. This a termed a List Comprehension.

image.png

```python
li=[1,2,3,4]
dli=[]
for i in li:
  dli.append(i*2)
print(dli)
```

```python
li=[1,2,3,4]
dli=[]
for i in li:
  dli=dli+[i*2]
print(dli)
```

```python
#wrt list comprehension
li =[1,2,3,4]
dli=[2*x for x in li]
print(dli)
```

```python
li=[1,2,3,4]
dli=[]
for i in li:
  dli.append(li*2)
print(dli)
```

```python
[2*i for i in [2,1,3]]
```

```python
[c=="a" for c in "Go away!"]
```

```python
squares = [i**2 for i in range(11)]
squares
```

List comprehensions with if condition:

```python
li =[3,2,1,0,-1,7,0.0]
dli=[2*x for x in li if x!=0]
dli
```

```python
nums =[3,1,4,1,5,9,2,6,5]
[num for num in nums if num%2==0]
```

```python
nums =[3,1,4,1,5,9,2,6,5]
[print("Even numbers printed",num) for num in nums if num%2==0]
```

```python
[str(x)+str(x) for x in range(13) if x%2==0]
```

```python
li=[['a',1],['b',2],['c',7]]
[n**2 for (x,n) in li]
```

```python
li=[3,6,2,7,1,9]
[elem*2 for elem in li if elem>4]
```
